<font size=16> Make noise file for forward modelling of WL

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
import healpy as hp
import numpy as np
import h5py, os

from msfm.utils import files, catalog
```

### Global constants

```python
conf = files.load_config( "../../../config_msfm_EuclidDR1F.yaml")

n_side = conf["analysis"]["n_side"]
n_pix = conf["analysis"]["n_pix"]
z_bins = conf["survey"]["WL"]["z_bins"]

# relative to this notebook
pixels_file = conf["files"]["pixels"]
noise_file = conf["files"]["noise"]
```

# Load relevant properties from Metacalibration catalog

```python
e1_tomo, e2_tomo, w_tomo, pix_tomo = catalog.get_shapes_from_cat(conf)
```

# Create the noise file

```python
with h5py.File(noise_file, "w") as f:
    for z_bin, e1, e2, w in zip(z_bins, e1_tomo, e2_tomo, w_tomo):
        print(f"{z_bin}: N_gal = {len(e1)}")
        
        # shape (n_gals, 3)
        sample = np.stack([e1, e2, w], axis=1)

        # save as np.float32, single precision
        dset = f.create_dataset(name=f"{z_bin}/cat", data=sample, dtype="f")
        dset.attrs["info"] = "This dataset contains the ellipticities of the WL galaxy sample per redshift bin. " \
                             "The shape is (n_gals, 3), where axis 1 contains (e1, e2, w)."
print(f'Stored {noise_file}')
```
